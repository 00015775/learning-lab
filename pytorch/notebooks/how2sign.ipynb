```python
from pathlib import Path
import hashlib

dirs = [
    Path("/content/extracted/test/frontal"),
    Path("/content/extracted/val/frontal")
]

def file_hash(path, chunk_size=8192):
    h = hashlib.md5()
    with open(path, "rb") as f:
        while chunk := f.read(chunk_size):
            h.update(chunk)
    return h.hexdigest()

unique_hashes = set()
total_files = 0

for d in dirs:
    files = list(d.rglob("*.npy"))
    total_files += len(files)

    for f in files:
        unique_hashes.add(file_hash(f))

    print(f"{d}: {len(files)} .npy files")

print(f"\nTotal .npy files (all): {total_files}")
print(f"Unique .npy files (by content): {len(unique_hashes)}")
```

```python
unique_names = set()

for d in dirs:
    for f in d.rglob("*.npy"):
        unique_names.add(f.name)

print(f"Unique .npy files (by name): {len(unique_names)}")
```

```python
from pathlib import Path
import numpy as np
import pandas as pd

# Directories
dirs = {
    "test": Path("/content/extracted/test/frontal"),
    "val": Path("/content/extracted/val/frontal")
}

# Expected shape pattern: last two dimensions
expected_last_two = (543, 3)

# Records for all files
records = []

for split, folder in dirs.items():
    for f in folder.rglob("*.npy"):
        try:
            arr = np.load(f)
            shape_ok = len(arr.shape) >= 2 and arr.shape[-2:] == expected_last_two
            nan_flag = np.isnan(arr).any()
            inf_flag = np.isinf(arr).any()

            status = "OK"
            if not shape_ok:
                status = "WRONG_SHAPE"
            if nan_flag:
                status = "HAS_NAN"
            if inf_flag:
                status = "HAS_INF"

            records.append({
                "split": split,
                "file": str(f),
                "shape": arr.shape,
                "dtype": arr.dtype,
                "min": arr.min(),
                "max": arr.max(),
                "mean": arr.mean(),
                "std": arr.std(),
                "nan": nan_flag,
                "inf": inf_flag,
                "status": status
            })
        except Exception as e:
            # Corrupted file
            records.append({
                "split": split,
                "file": str(f),
                "shape": None,
                "dtype": None,
                "min": None,
                "max": None,
                "mean": None,
                "std": None,
                "nan": None,
                "inf": None,
                "status": f"CORRUPTED: {e}"
            })

# Convert to DataFrame
df = pd.DataFrame(records)

# Display first few rows
display(df)

# Summary statistics
summary = df['status'].value_counts()
print("Status summary across both splits:")
print(summary)

# Optional: check per split
split_summary = df.groupby('split')['status'].value_counts()
print("\nStatus per split:")
print(split_summary)
```

### global stats per split

```python
split_info = {}
for split, folder in dirs.items():
    global_max_arr = None  # shape (X, 543, 3)
    global_min_arr = None
    global_max_value = -np.inf
    global_min_value = np.inf
    nan_flag = False
    inf_flag = False
    corrupted_flag = False

    for f in folder.rglob("*.npy"):
        try:
            arr = np.load(f)
            if arr.ndim != 3 or arr.shape[-2:] != expected_last_two:
                continue
            if np.isnan(arr).any():
                nan_flag = True
            if np.isinf(arr).any():
                inf_flag = True

            file_max_value = arr.max()
            file_min_value = arr.min()

            if file_max_value > global_max_value:
                global_max_value = file_max_value
                global_max_arr = arr  # store full array (X, 543, 3)

            if file_min_value < global_min_value:
                global_min_value = file_min_value
                global_min_arr = arr  # store full array (X, 543, 3)

        except Exception as e:
            corrupted_flag = True

    split_info[split] = {
        "global_max_arr": global_max_arr,
        "global_min_arr": global_min_arr,
        "global_max_value": global_max_value,
        "global_min_value": global_min_value,
        "any_nan": nan_flag,
        "any_inf": inf_flag,
        "any_corrupted": corrupted_flag
    }

# Display summary
for split, info in split_info.items():
    print(f"\n=== Split: {split} ===")
    print(f"Global max value: {info['global_max_value']}")
    print(f"Global min value: {info['global_min_value']}")
    print(f"Any NaN values: {info['any_nan']}")
    print(f"Any Inf values: {info['any_inf']}")
    print(f"Any corrupted files: {info['any_corrupted']}")
    if info['global_max_arr'] is not None:
        print(f"Global max array shape: {info['global_max_arr'].shape}")
        print(f"Global min array shape: {info['global_min_arr'].shape}")
```
